```python
def get_user_list(size=20):
    """Accepts 'size' integer values from the user and returns a list."""
    values = []
    print(f"Enter {size} integer values:")
    for i in range(size):
        while True:
            try:
                num = int(input(f"Value {i+1}: "))
                values.append(num)
                break
            except ValueError:
                print("Invalid input! Please enter an integer.")
    return values

def find_similar_elements(lst):
    """Finds elements that occur more than once and prints their indexes."""
    seen = {}
    for idx, val in enumerate(lst):
        seen.setdefault(val, []).append(idx)
    
    print("\nSimilar elements and their indexes:")
    found = False
    for val, indexes in seen.items():
        if len(indexes) > 1:
            print(f"Value {val} found at indexes {indexes}")
            found = True
    if not found:
        print("No similar elements found.")

def count_even_odd(lst):
    """Counts even and odd numbers in the list."""
    even_count = sum(1 for x in lst if x % 2 == 0)
    odd_count = len(lst) - even_count
    print(f"\nEven numbers: {even_count}")
    print(f"Odd numbers: {odd_count}")

def count_positive_negative(lst):
    """Counts positive and negative numbers in the list."""
    positive_count = sum(1 for x in lst if x > 0)
    negative_count = sum(1 for x in lst if x < 0)
    print(f"\nPositive numbers: {positive_count}")
    print(f"Negative numbers: {negative_count}")

# Main execution
if __name__ == "__main__":
    numbers = get_user_list(20)
    print("\nList entered:", numbers)
    
    find_similar_elements(numbers)
    count_even_odd(numbers)
    count_positive_negative(numbers)
```
